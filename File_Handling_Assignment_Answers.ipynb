# Part A — Theory Questions

## 1. What is the difference between interpreted and compiled languages?

Compiled languages are translated into machine code before execution; interpreted languages execute through an interpreter or runtime. Python normally compiles source into bytecode, then runs it on a virtual machine.

## 2. What is exception handling in Python?

Exception handling lets a program detect and respond to runtime errors without terminating unexpectedly. Python uses try, except, else, and finally.

## 3. What is the purpose of the finally block in exception handling?

The finally block runs whether an exception occurs or not, and is commonly used for cleanup such as closing resources.

## 4. What is logging in Python?

Logging records program events, warnings, and errors to the console or a file for debugging and monitoring.

## 5. What is the significance of the __del__ method in Python?

__del__ is a finalizer that may run when an object is collected. Its timing is not guaranteed; use context managers for reliable resource cleanup.

## 6. What is the difference between import and from ... import in Python?

import math imports a module and accesses math.sqrt(9); from math import sqrt imports the name directly and allows sqrt(9).

## 7. How can you handle multiple exceptions in Python?

Use separate except clauses for different exception types, or a tuple such as except (ValueError, TypeError) as error.

## 8. What is the purpose of the with statement when handling files in Python?

The with statement uses a context manager to close a file automatically even when an exception occurs.

## 9. What is the difference between multithreading and multiprocessing?

Multithreading runs threads within one process and shares memory; multiprocessing uses separate processes with separate memory. Processes can achieve CPU parallelism in standard CPython.

## 10. What are the advantages of using logging in a program?

Logging provides timestamps, severity levels, persistent records, easier troubleshooting, and production monitoring.

## 11. What is memory management in Python?

Memory management is the allocation, tracking, and release of memory for objects while a Python program runs.

## 12. What are the basic steps involved in exception handling in Python?

Place risky code in try, catch expected errors in except, optionally use else for success-only code, and finally for cleanup.

## 13. Why is memory management important in Python?

Good memory management prevents excessive memory use, improves performance, and reduces crashes and resource leaks.

## 14. What is the role of try and except in exception handling?

Try encloses code that may fail; except catches specified exceptions and performs a recovery or reporting action.

## 15. How does Python's garbage collection system work?

In CPython, reference counting reclaims most unreachable objects promptly, while cyclic garbage collection detects reference cycles.

## 16. What is the purpose of the else block in exception handling?

An else block runs only when the try block completes without raising an exception.

## 17. What are the common logging levels in Python?

DEBUG, INFO, WARNING, ERROR, and CRITICAL are standard logging levels, from detailed diagnostics to severe failures.

## 18. What is the difference between os.fork() and multiprocessing in Python?

os.fork() duplicates the current process on supported Unix-like systems. multiprocessing offers a higher-level, cross-platform API for processes, communication, and synchronization.

## 19. What is the importance of closing a file in Python?

Closing files flushes buffered writes and releases operating-system resources; with open(...) is recommended.

## 20. What is the difference between file.read() and file.readline() in Python?

file.read() reads all remaining content (or a specified number of characters); file.readline() reads one line at a time.

## 21. What is the logging module in Python used for?

The logging module provides configurable loggers, handlers, formatters, and severity levels to record application activity.

## 22. What is the os module in Python used for in file handling?

The os module helps work with file paths, directories, file metadata, renaming, deletion, and operating-system operations.

## 23. What are the challenges associated with memory management in Python?

Challenges include reference cycles, lingering references, large object allocations, fragmentation, and native-library memory usage.

## 24. How do you raise an exception manually in Python?

Use raise ValueError("Invalid value") to raise an exception manually.

## 25. Why is it important to use multithreading in certain applications?

Multithreading is useful for concurrent I/O tasks such as downloads, network requests, and waiting for files, improving responsiveness.

## 3. How would you handle a case where the file doesn't exist while trying to open it for reading?

```python
try:
    with open("missing.txt", encoding="utf-8") as file:
        print(file.read())
except FileNotFoundError:
    print("The file does not exist.")
```

## 4. Write a Python script that reads from one file and writes its content to another file.

```python
with open("source.txt", "r", encoding="utf-8") as src:
    with open("copy.txt", "w", encoding="utf-8") as dst:
        dst.write(src.read())
```

## 5. How would you catch and handle division by zero error in Python?

```python
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
import logging
logging.basicConfig(filename="errors.log", level=logging.ERROR)
try:
    result = 10 / 0
except ZeroDivisionError:
    logging.exception("Division by zero occurred")
```

## 16. Demonstrate how to use memory profiling to check the memory usage of a small program.

```python
# Standard-library memory measurement (Python allocations)
import tracemalloc
tracemalloc.start()
numbers = [i * i for i in range(10000)]
current, peak = tracemalloc.get_traced_memory()
print(f"Current: {current / 1024:.1f} KiB")
print(f"Peak: {peak / 1024:.1f} KiB")
tracemalloc.stop()
```

## 18. How would you implement a basic logging setup that logs to a file with rotation after 1MB?

```python
import logging
from logging.handlers import RotatingFileHandler
logger = logging.getLogger("rotating_example")
logger.setLevel(logging.INFO)
handler = RotatingFileHandler("app.log", maxBytes=1024 * 1024, backupCount=3)
logger.addHandler(handler)
logger.info("Log rotation is configured")
```

## 21. Write a Python program that reads a file and prints the number of occurrences of a specific word.

```python
import re
with open("example.txt", encoding="utf-8") as file:
    content = file.read()
word = "python"
count = sum(token.casefold() == word.casefold()
            for token in re.findall(r"\b\w+\b", content))
print(f"Occurrences of {word}: {count}")
```

## 22. How can you check if a file is empty before attempting to read its contents?

```python
from pathlib import Path
path = Path("example.txt")
if path.is_file() and path.stat().st_size == 0:
    print("File is empty")
elif path.is_file():
    print("File has content")
else:
    print("File does not exist")
```
