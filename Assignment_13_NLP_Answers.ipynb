Question 1: Natural Language Processing and its applications

Natural Language Processing (NLP) is a branch of artificial intelligence that enables computers to process, analyze, understand, and generate human language in text or speech form.

Machine translation: translating sentences between languages, such as English and Hindi.

Sentiment analysis: classifying customer reviews as positive, negative, or neutral.

Chatbots and virtual assistants: answering questions and responding to spoken or typed commands.

Spam detection: classifying email messages as spam or legitimate.

Computers cannot directly interpret raw text as meaning: text contains ambiguous words, grammatical structure, context, variations in spelling, and nonnumeric symbols. Algorithms generally need a numerical representation of text.

Basic NLP pipeline: (1) collect text; (2) clean and normalize it; (3) tokenize into words or subwords; (4) extract features such as BoW, TF-IDF, or embeddings; (5) train or apply an NLP model; (6) evaluate predictions and deploy the solution.

Question 2: Text preprocessing operations

Text preprocessing makes noisy, inconsistent text easier to analyze and reduces irrelevant variation. Different tasks may require different preprocessing choices.

Lowercasing: "Hello WORLD" → "hello world".

Removing punctuation: "Hello, world!" → "Hello world".

Removing numbers: "Order 123 arrived" → "Order arrived" (only if digits are not important).

Removing stopwords: "This is a good book" → "good book" (depending on the stopword list).

Handling extra whitespaces: "Hello    world" → "Hello world".

Tokenization: "I like NLP" → ["I", "like", "NLP"].

Stemming: "studies", "studying" → stems such as "studi" (Porter stemmer).

Lemmatization: "running" → "run" when analyzed as a verb; "better" → "good" with appropriate POS information.

N-grams: for "I love NLP", unigrams = ["I", "love", "NLP"]; bigrams = ["I love", "love NLP"]; trigram = ["I love NLP"].

Question 3: NLTK, spaCy, Gensim and word embeddings

NLTK: education, linguistic research, tokenization, stemming, corpora, and classic NLP experiments.

spaCy: fast production-oriented pipelines for tokenization, POS tagging, dependency parsing, lemmatization, and named entity recognition.

Gensim: topic modeling and vector representations, especially Word2Vec, Doc2Vec, and LDA.

Word embeddings represent words as dense numerical vectors learned from language data. Words occurring in similar contexts can have nearby vectors.

Bag of Words (BoW) uses sparse word-count vectors and ignores word order. TF-IDF uses sparse vectors weighted by term frequency and document rarity. Dense embeddings have far fewer dimensions than a typical vocabulary-sized vector, can capture semantic similarity, and are learned rather than directly counted. Static embeddings such as Word2Vec give one vector per word; contextual embeddings can vary by sentence.

Question 4: Sequential data, RNN, LSTM, GRU and evaluation

Sequential data consists of ordered observations, such as the words of a sentence. Their order and context affect meaning. A standard feed-forward network processing independent fixed-size inputs does not inherently maintain memory of previous tokens or handle varying sequence lengths; recurrent models use state across time steps.

RNN: updates a hidden state at each step; simple but susceptible to vanishing or exploding gradients on long sequences.

LSTM: uses input, forget, and output gates plus a cell state to preserve useful long-term information; more parameters.

GRU: uses update and reset gates, with no separate cell state; often simpler and faster than LSTM.

For binary classification, TP = true positives, TN = true negatives, FP = false positives, FN = false negatives.

Accuracy = (TP + TN) / (TP + TN + FP + FN).

Precision = TP / (TP + FP): fraction of predicted positives that are correct.

Recall = TP / (TP + FN): fraction of actual positives detected.

F1-score = 2 × Precision × Recall / (Precision + Recall): harmonic mean of precision and recall.

Perplexity = exp(−(1/N) × sum of log probabilities of the N correct next tokens); lower perplexity indicates better next-token prediction on the same evaluation data.

Question 5: Sequence-to-Sequence and attention

A sequence-to-sequence (Seq2Seq) model transforms an input sequence into an output sequence, e.g., for translation or summarization. An encoder reads input tokens and produces hidden states. A decoder generates output tokens one by one, conditioned on encoded information and previously generated tokens. In a basic encoder–decoder, the context vector is commonly the encoder’s final hidden representation passed to the decoder.

Basic Seq2Seq compresses a long input into a fixed-size context vector, causing an information bottleneck and difficulty retaining distant details. Attention computes relevance scores between the decoder’s current state and all encoder hidden states; softmax converts these scores into weights. A weighted sum creates a context vector for each decoding step, letting the decoder focus on different source positions.

Question 6: Reusable text-preprocessing function

The following code lowercases, removes punctuation and numbers, normalizes spaces, removes English stopwords, and returns tokens.

```python
import re
import string
import nltk
from nltk.corpus import stopwords
 
nltk.download("stopwords", quiet=True)
STOPWORDS = set(stopwords.words("english"))
 
def preprocess_text(text):
    text = text.lower()
    text = text.translate(str.maketrans("", "", string.punctuation))
    text = re.sub(r"\d+", "", text)
    text = " ".join(text.split())
    tokens = text.split()
    return [word for word in tokens if word not in STOPWORDS]
 
sample = " This PRODUCT is Amazing!!! I bought 2 items, and I really loved them. "
print(preprocess_text(sample))
```

Expected output with the standard NLTK English stopword list: ['product', 'amazing', 'bought', 'items', 'really', 'loved'].

Question 7: NLTK tokenization, stemming, lemmatization and n-grams

```python
import nltk
from nltk.tokenize import TreebankWordTokenizer
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.util import ngrams
 
nltk.download("wordnet", quiet=True)
sentence = "The students were studying and running towards their classes."
tokens = TreebankWordTokenizer().tokenize(sentence)
words = [t for t in tokens if t.isalpha()]
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
print("Tokens:", tokens)
print("Stems:", [stemmer.stem(w) for w in words])
print("Lemmas (default noun):", [lemmatizer.lemmatize(w) for w in words])
print("Lemmas (verb for studying/running):",
      [lemmatizer.lemmatize(w, pos="v") for w in words])
for n in (1, 2, 3):
    print(f"{n}-grams:", list(ngrams(words, n)))
```

Comparison: Porter stemming can produce non-dictionary roots (e.g., "studi" for "studying"); WordNet lemmatization aims for valid base forms ("study" for "studying" with verb POS). Lemmatization depends on the supplied part-of-speech tag; default noun lemmatization may leave verbs unchanged.

Question 8: Bag of Words and TF-IDF

```python
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
 
docs = [
    "NLP is useful for text analysis",
    "Text analysis is useful for business",
    "NLP converts text into numerical features"
]
bow = CountVectorizer()
counts = bow.fit_transform(docs)
print("Vocabulary:", bow.get_feature_names_out().tolist())
print("BoW matrix:\n", counts.toarray())
 
tfidf = TfidfVectorizer()
values = tfidf.fit_transform(docs).toarray()
terms = tfidf.get_feature_names_out()
print("TF-IDF matrix (rounded):\n", np.round(values, 3))
for i, row in enumerate(values, start=1):
    highest = np.flatnonzero(np.isclose(row, row.max()))
    print(f"Document {i} highest TF-IDF:", terms[highest].tolist(),
          "score:", round(row.max(), 3))
```

Vocabulary (alphabetical): analysis, business, converts, features, for, into, is, nlp, numerical, text, useful. CountVectorizer uses lowercase tokens by default. With default smoothed IDF and L2 normalization, the maximum score may be tied across multiple words; the program reports all ties.

Question 9: spaCy analysis and Gensim Word2Vec

Part A — Install the packages and English model before running: pip install spacy gensim; python -m spacy download en_core_web_sm.

```python
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Microsoft opened a new AI research centre in Bengaluru in 2025."
doc = nlp(text)
for token in doc:
    print(token.text, "| lemma:", token.lemma_, "| POS:", token.pos_)
print("Named entities:", [(ent.text, ent.label_) for ent in doc.ents])
```

The trained model typically identifies Microsoft as ORG, Bengaluru as GPE, and 2025 as DATE; exact tagging can differ by model version.

Part B — A small demonstration corpus; nearest neighbors are illustrative and depend on training.

```python
from gensim.models import Word2Vec
sentences = [
    ["king", "queen", "royal", "palace"],
    ["king", "prince", "royal", "throne"],
    ["queen", "princess", "royal", "palace"],
    ["man", "king", "throne", "palace"],
    ["woman", "queen", "palace", "royal"]
] * 100
model = Word2Vec(sentences, vector_size=30, window=3,
                 min_count=1, workers=1, epochs=100, seed=42)
print(model.wv.most_similar("king", topn=5))
```

Because this corpus is tiny, the similarities are not reliable measures of real-world semantic relationships.

Question 10: LSTM sentiment model and attention weights

Part A — Example binary sentiment classifier using TensorFlow/Keras. It is a working demonstration with a tiny toy dataset, not a production-quality classifier.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Sequential
 
texts = np.array([
    "I loved this movie", "This was excellent", "Very enjoyable film",
    "Amazing story and acting", "I hated this movie", "This was terrible",
    "Very boring film", "Awful story and acting"
])
labels = np.array([1, 1, 1, 1, 0, 0, 0, 0])
vectorizer = layers.TextVectorization(max_tokens=1000,
                                      output_sequence_length=12)
vectorizer.adapt(texts)
model = Sequential([
    layers.Input(shape=(), dtype=tf.string),
    vectorizer,
    layers.Embedding(input_dim=1000, output_dim=16, mask_zero=True),
    layers.LSTM(16),
    layers.Dense(1, activation="sigmoid")
])
model.compile(optimizer="adam", loss="binary_crossentropy",
              metrics=["accuracy"])
model.fit(texts, labels, epochs=15, verbose=0)
print(model.predict(np.array(["I loved the acting"]), verbose=0))
```
