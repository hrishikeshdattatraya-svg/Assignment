## Question 1

> What is the difference between descriptive statistics and inferential statistics? Explain with examples.

### Answer

**Descriptive statistics** summarize and describe the data that has actually been collected. **Inferential statistics** use a sample of data to draw conclusions, make estimates or test claims about a larger population that was not fully measured.

| Aspect | Descriptive statistics | Inferential statistics |
|---|---|---|
| Purpose | Describe and summarize the data at hand | Generalize from a sample to a population and make predictions or decisions |
| Data used | The whole group being studied (or the given data set) | A sample taken from a larger population |
| Main tools | Mean, median, mode, range, standard deviation, tables, bar charts, histograms, boxplots | Hypothesis tests (t-test, chi-square), confidence intervals, regression, ANOVA |
| Result | Facts about the data (no uncertainty) | Conclusions with a degree of uncertainty (p-values, confidence levels) |
| Question answered | "What does this data look like?" | "What can we conclude about the population?" |

**Examples**

- **Descriptive:** the marks of all 50 students in a class are collected. The average mark is 72, the highest is 95, and a histogram shows how the marks are spread. Nothing is said about students outside this class.
- **Inferential:** a survey of 200 randomly selected voters shows that 54% support a candidate. Using a confidence interval (for example 54% ± 7%), we estimate the support among all voters in the city. Testing whether a new teaching method really improves marks compared with the old one is also inferential.

## Question 2

> What is sampling in statistics? Explain the differences between random and stratified sampling.

### Answer

**Sampling** is the process of selecting a smaller group (the **sample**) from a larger group (the **population**) so that conclusions about the population can be drawn by studying only the sample. It is used because studying every member is usually too costly, too slow or impossible. A good sample must be representative of the population.

**Random sampling (simple random sampling):** every member of the population has an equal chance of being selected, and the selection is done purely by chance, for example by a lottery or a random number generator. Example: choosing 100 students at random from a list of 1,000 students.

**Stratified sampling:** the population is first divided into non-overlapping, similar groups called **strata** (for example by gender, age group or region). A random sample is then drawn from each stratum, usually in proportion to its size. Example: in a school of 1,000 students with 600 boys and 400 girls, a sample of 100 is formed by randomly choosing 60 boys and 40 girls.

| Aspect | Random sampling | Stratified sampling |
|---|---|---|
| Method | Select members directly from the whole population by chance | Divide into strata, then select randomly within each stratum |
| Chance of selection | Equal for everyone | Equal within each stratum |
| Representation | Subgroups may be over- or under-represented by chance | Every subgroup is guaranteed to be represented |
| Precision | Lower when the population is very varied | Usually higher (smaller sampling error) for the same sample size |
| Requirement | A complete list of the population | Also needs information to form the strata |
| Best used when | The population is fairly uniform | The population has distinct subgroups that matter |

## Question 3

> Define mean, median, and mode. Explain why these measures of central tendency are important.

### Answer

- **Mean (average):** the sum of all values divided by the number of values, mean = Σx / n. It uses every value in the data.
- **Median:** the middle value when the data is arranged in order. For an even number of values it is the average of the two middle values. It is not affected by extreme values.
- **Mode:** the value that occurs most often. A data set can have one mode, more than one mode, or none. It also works for categorical data.

Example: for the data 2, 3, 3, 5, 7 the mean is 20 / 5 = 4, the median is 3 (the middle value) and the mode is 3 (occurs twice).

**Why they are important**

- They give a **single representative value** that summarizes a large data set.
- They make it easy to **compare** different groups or time periods (for example average sales of two stores).
- They show the **centre** of the distribution and, when compared with each other, hint at its shape (mean greater than median suggests a right skew).
- Each suits a different situation: the mean for symmetric data, the median for skewed data or data with outliers (such as income), and the mode for the most common category or size (such as the most popular shoe size).
- They are the starting point for other statistics such as variance, standard deviation and hypothesis tests.

## Question 4

> Explain skewness and kurtosis. What does a positive skew imply about the data?

### Answer

**Skewness** measures the **asymmetry** of a distribution around its mean, that is, whether one tail is longer than the other.

| Type of skew | Shape | Relationship of averages | Example |
|---|---|---|---|
| Positive (right) skew | Long tail on the right | Mean > Median > Mode | Income, house prices |
| Zero skew (symmetric) | Both sides are mirror images | Mean = Median = Mode | Heights, normal distribution |
| Negative (left) skew | Long tail on the left | Mean < Median < Mode | Exam scores on an easy test |

**Kurtosis** measures the **tailedness** of a distribution, that is, how heavy its tails are (how prone it is to extreme values) compared with a normal distribution. The normal distribution has a kurtosis of 3, which is an excess kurtosis of 0.

- **Mesokurtic:** excess kurtosis about 0, tails similar to a normal distribution.
- **Leptokurtic:** excess kurtosis above 0, a sharper peak and heavy tails, so more outliers.
- **Platykurtic:** excess kurtosis below 0, a flatter shape and light tails, so fewer outliers.

**What a positive skew implies**

- Most values are clustered at the lower end, and a few unusually **large values** stretch the right tail.
- The **mean is pulled above the median**, so the mean overstates the "typical" value. The median is then a better measure of centre.
- Examples are income, property prices and waiting times. Analysts often check for outliers or apply a transformation such as the logarithm to reduce the skew.

## Question 5

> Implement a Python program to compute the mean, median, and mode of a given list of numbers.
>
> `numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]`

### Answer

```python
from collections import Counter
import statistics

numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

# Mean = sum of all values / number of values
mean_value = sum(numbers) / len(numbers)

# Median = middle value of the sorted list
sorted_numbers = sorted(numbers)
n = len(sorted_numbers)
mid = n // 2
if n % 2 == 1:
    median_value = sorted_numbers[mid]
else:
    median_value = (sorted_numbers[mid - 1] + sorted_numbers[mid]) / 2

# Mode = most frequent value(s); there can be more than one
counts = Counter(numbers)
highest = max(counts.values())
mode_values = sorted(value for value, c in counts.items() if c == highest)

print("Numbers :", numbers)
print("Mean    :", mean_value)
print("Median  :", median_value)
print("Mode    :", mode_values, "(each occurs", highest, "times)")

# Cross-check with Python's statistics module
print("\nCheck with the statistics module:")
print("mean =", statistics.mean(numbers))
print("median =", statistics.median(numbers))
print("multimode =", statistics.multimode(numbers))
```

**Explanation:** the 15 numbers add up to 294, so the mean is 294 / 15 = **19.6**. After sorting, the middle (8th) value is 19, so the median is **19**. The values 12, 19 and 24 each occur three times, which is more than any other value, so the data is **multimodal** with modes **12, 19 and 24**. (`statistics.mode()` would return only the first mode, 12, which is why `multimode()` or the counting method is used.)

## Question 6

> Compute the covariance and correlation coefficient between the following two datasets provided as lists in Python:
>
> `list_x = [10, 20, 30, 40, 50]`
>
> `list_y = [15, 25, 35, 45, 60]`

### Answer

```python
import numpy as np

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

# np.cov gives a 2x2 covariance matrix; [0, 1] is cov(x, y)
sample_cov = np.cov(list_x, list_y)[0, 1]                 # divides by n - 1
population_cov = np.cov(list_x, list_y, bias=True)[0, 1]  # divides by n

# np.corrcoef gives a 2x2 correlation matrix; [0, 1] is r
correlation = np.corrcoef(list_x, list_y)[0, 1]

print("Sample covariance     :", sample_cov)
print("Population covariance :", population_cov)
print("Correlation (Pearson) :", round(correlation, 4))
```

**Explanation:** the means are 30 (x) and 36 (y). The covariance is positive (275 for a sample, 220 for a population), so y tends to rise when x rises. Covariance depends on the units, so its size alone is hard to judge. The correlation coefficient scales it to a range from -1 to +1: **r ≈ 0.9959**, a **very strong positive linear relationship**. The pair is not perfectly correlated only because the last y value (60) is slightly above the straight-line pattern (55).

## Question 7

> Write a Python script to draw a boxplot for the following numeric list and identify its outliers. Explain the result:
>
> `data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]`

### Answer

```python
import numpy as np
import matplotlib.pyplot as plt

data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

# Outliers using the IQR rule
q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr
outliers = [x for x in data if x < lower_fence or x > upper_fence]

print("Q1 (25th percentile):", q1)
print("Q3 (75th percentile):", q3)
print("IQR                 :", iqr)
print("Lower fence         :", lower_fence)
print("Upper fence         :", upper_fence)
print("Outliers            :", outliers)

# Boxplot
plt.figure(figsize=(6, 6))
plt.boxplot(data, patch_artist=True, boxprops=dict(facecolor="#9ecae1"))
plt.title("Boxplot of the Data")
plt.ylabel("Value")
plt.show()
```

**Explanation of the result**

- The box runs from Q1 = 17.25 to Q3 = 23.25, so the **interquartile range (IQR) is 6**. The line inside the box is the median (21.5).
- Outliers are values beyond 1.5 × IQR from the box. The **lower fence is 8.25** and the **upper fence is 32.25**.
- No value is below 8.25, and the only value above 32.25 is **35**, so **35 is the single outlier**. It is drawn as a separate circle.
- The whiskers stop at the smallest and largest values inside the fences: 12 and 29.
- The outlier lies far above the rest of the data on the upper side, which pulls the distribution slightly to the right. Before removing it, check whether it is a data-entry error or a genuine extreme value.

## Question 8

> You are working as a data analyst in an e-commerce company. The marketing team wants to know if there is a relationship between advertising spend and daily sales.
>
> - Explain how you would use covariance and correlation to explore this relationship.
> - Write Python code to compute the correlation between the two lists:
>
> `advertising_spend = [200, 250, 300, 400, 500]`
>
> `daily_sales = [2200, 2450, 2750, 3200, 4000]`

### Answer

**(a) Using covariance and correlation**

- **Collect paired data:** advertising spend and sales for the same days, and first draw a scatter plot to see whether the points form a pattern.
- **Covariance** shows the **direction** of the relationship. A positive value means sales tend to rise when spend rises, a negative value means they fall, and a value near zero means no linear pattern. Its size depends on the units (rupees or dollars), so it cannot show how strong the relationship is.
- **Correlation coefficient (r)** is the covariance divided by the product of the two standard deviations. It is unit-free and lies between -1 and +1, so it shows both direction and **strength** (close to +1 or -1 is strong, close to 0 is weak).
- **Caution:** correlation does not prove causation. Seasonality, discounts or festivals may raise both spend and sales. Correlation only measures linear relationships and is sensitive to outliers, and a small sample gives unreliable results. A regression analysis or a controlled test would be the next step.

**(b) Python code and output**

```python
import numpy as np

advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

covariance = np.cov(advertising_spend, daily_sales)[0, 1]
correlation = np.corrcoef(advertising_spend, daily_sales)[0, 1]

print("Covariance (sample):", covariance)
print("Correlation (r)    :", round(correlation, 4))
```

**Interpretation:** the covariance is positive (84,875) and the correlation is **r ≈ 0.9936**, a very strong positive linear relationship: days with higher advertising spend have higher sales. r² is about 0.987, meaning roughly 98.7% of the variation in sales in this sample is linearly associated with advertising spend. Because there are only 5 data points and correlation is not causation, the marketing team should confirm this with more data before concluding that advertising causes the increase in sales.

## Question 9

> Your team has collected customer satisfaction survey data on a scale of 1-10 and wants to understand its distribution before launching a new product.
>
> - Explain which summary statistics and visualizations (e.g. mean, standard deviation, histogram) you'd use.
> - Write Python code to create a histogram using Matplotlib for the survey data:
>
> `survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]`

### Answer

**(a) Summary statistics and visualizations**

- **Central tendency:** mean, median and mode, to see what a typical customer scored.
- **Spread:** standard deviation (or variance), range and quartiles / IQR, to see how much opinions differ.
- **Shape:** skewness, to see whether scores bunch up at the high or low end.
- **Frequency table and percentages:** for example the share of customers who gave 8 or more.
- **Histogram or bar chart:** shows the shape of the distribution and where scores are concentrated.
- **Boxplot:** shows the median, spread and any outliers, and is useful for comparing groups such as regions.

**(b) Python code and output**

```python
import numpy as np
import matplotlib.pyplot as plt

survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

# Summary statistics
mean_score = np.mean(survey_scores)
median_score = np.median(survey_scores)
std_score = np.std(survey_scores, ddof=1)      # sample standard deviation

print("Mean              :", round(mean_score, 2))
print("Median            :", median_score)
print("Standard deviation:", round(std_score, 2))
print("Minimum / Maximum :", min(survey_scores), "/", max(survey_scores))

# Histogram (one bar for each score from 1 to 10)
plt.figure(figsize=(8, 5))
plt.hist(survey_scores, bins=range(1, 12), align="left",
         edgecolor="black", color="#4C78A8")
plt.axvline(mean_score, color="red", linestyle="--",
            label=f"Mean = {mean_score:.2f}")
plt.xticks(range(1, 11))
plt.yticks(range(0, 5))
plt.xlabel("Satisfaction score (1-10)")
plt.ylabel("Number of customers")
plt.title("Distribution of Customer Satisfaction Scores")
plt.legend()
plt.show()
```

**Interpretation:** the average score is 7.33 with a median of 7 and a sample standard deviation of 1.63, and scores range from 4 to 10. The most common score is 7 (4 customers). About 73% of customers (11 of 15) scored 7 or higher, and nobody scored below 4, so customers are generally satisfied. The distribution has one peak around 7 to 9 and is fairly balanced on both sides, so the mean and median are close.
