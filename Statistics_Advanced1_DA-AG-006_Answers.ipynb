## Question 1

> What is a random variable in probability theory?

### Answer

A **random variable** is a variable whose value is a numerical outcome of a random experiment. Formally, it is a function that assigns a real number to every possible outcome in the sample space. It is usually written with a capital letter such as X, and its particular values with a small letter such as x.

Its importance is that it turns outcomes such as "heads" or "defective" into numbers, so that probabilities can be calculated and summarized with a **probability distribution**, an expected value and a variance.

**Examples**

- Tossing a coin 3 times: let X = the number of heads. X can take the values 0, 1, 2 or 3.
- Rolling a die: let X = the number shown. X can take the values 1, 2, 3, 4, 5 or 6, each with probability 1/6.
- Measuring the height of a randomly chosen person: X is a random variable that can take any value in a range of heights.

Random variables are of two main types, **discrete** and **continuous** (see Question 2).

## Question 2

> What are the types of random variables?

### Answer

There are two main types of random variables.

| Aspect | Discrete random variable | Continuous random variable |
|---|---|---|
| Values | Countable values (usually whole numbers), with gaps between them | Any value in an interval; uncountably many possible values |
| Comes from | Counting | Measuring |
| Probability function | Probability mass function (PMF), P(X = x) | Probability density function (PDF), f(x) |
| Probability of one exact value | Can be greater than 0 | Always 0; probabilities are found for intervals |
| Examples | Number of heads in 5 tosses, number of defective items, number of customers in a day | Height, weight, temperature, time to complete a task |

Some situations also produce a **mixed** random variable, which has both discrete and continuous parts (for example, the waiting time at a counter that is exactly 0 for many people and a continuous value for the others).

## Question 3

> Explain the difference between discrete and continuous distributions.

### Answer

A **probability distribution** describes how the probabilities are spread over the possible values of a random variable. A **discrete distribution** belongs to a discrete random variable, and a **continuous distribution** belongs to a continuous random variable.

| Feature | Discrete distribution | Continuous distribution |
|---|---|---|
| Describes | Probability of each exact value | Probability over ranges (intervals) of values |
| Function | PMF: P(X = x) | PDF: f(x) (density, not a probability) |
| P(X = x) | Can be positive | Equal to 0 for any single point |
| Total probability | Sum of all probabilities = 1 | Total area under the curve = 1 |
| Finding P(a ≤ X ≤ b) | Add the probabilities of the values | Integrate the PDF from a to b (area under the curve) |
| CDF shape | Step function | Smooth continuous curve |
| Graph | Bar chart / spikes | Smooth curve |
| Common examples | Bernoulli, Binomial, Poisson, Geometric | Normal, Uniform, Exponential, t-distribution |

## Question 4

> What is a binomial distribution, and how is it used in probability?

### Answer

The **binomial distribution** gives the probability of getting exactly k successes in n independent trials of an experiment, when each trial has only two outcomes (success or failure) and the probability of success p is the same in every trial. It is written X ~ Binomial(n, p).

**Conditions**

- A fixed number of trials, n.
- Each trial has only two outcomes: success or failure.
- The trials are independent of each other.
- The probability of success, p, is constant for each trial.

**Formula**

$$P(X = k) = \binom{n}{k}\, p^{k}\,(1-p)^{n-k}, \qquad k = 0, 1, 2, \dots, n$$

where $\binom{n}{k} = \dfrac{n!}{k!\,(n-k)!}$

$$\text{Mean} = np \qquad\qquad \text{Variance} = np(1-p)$$

**Example:** a fair coin is tossed 10 times. The probability of exactly 6 heads is C(10, 6) × 0.5⁶ × 0.5⁴ = 210 / 1024 ≈ **0.2051**.

**Uses**

- Quality control: the number of defective items in a batch of n items.
- Surveys and polls: the number of people out of n who prefer a product.
- Medical trials: the number of patients out of n who respond to a treatment.
- Games of chance and A/B testing (number of conversions out of n visitors).

## Question 5

> What is the standard normal distribution, and why is it important?

### Answer

The **standard normal distribution** is the normal distribution with **mean 0 and standard deviation 1**, written Z ~ N(0, 1). Its curve is a symmetric bell shape centred at 0, and the total area under it is 1.

Any normal variable X with mean μ and standard deviation σ can be converted to the standard normal by the **z-score** formula:

$$z = \frac{X - \mu}{\sigma}$$

**Example:** an exam score of 70 in a class with mean 60 and standard deviation 5 has z = (70 - 60) / 5 = 2, that is, two standard deviations above the mean.

**Why it is important**

- **Standardization:** values measured in different units or scales can be compared using z-scores.
- **Probability tables:** areas under the standard normal curve are tabulated (z-table) and built into software, so probabilities for any normal distribution can be found by converting to z.
- **Empirical rule:** about 68% of values lie within 1, about 95% within 2 and about 99.7% within 3 standard deviations of the mean.
- **Inference:** confidence intervals and hypothesis tests use standard normal values, for example z = 1.96 for a 95% confidence level.
- **Outlier detection:** values with |z| greater than 3 are usually treated as unusual.
- It is the limiting distribution described by the Central Limit Theorem.

## Question 6

> What is the Central Limit Theorem (CLT), and why is it critical in statistics?

### Answer

The **Central Limit Theorem** states that if independent random samples of size n are drawn from any population with a finite mean μ and finite standard deviation σ, then the distribution of the **sample mean** becomes approximately normal as n grows large, no matter what the shape of the original population is. The sampling distribution of the mean has:

- mean = μ (the population mean), and
- standard deviation = σ / √n, called the **standard error**.

A common rule of thumb is that n ≥ 30 is large enough. Very skewed populations may need a larger n, and populations that are already close to normal need a smaller one.

**Example:** the value of a single die roll is uniformly distributed (not bell-shaped). But if we roll a die 30 times and record the average, and repeat this many times, the averages form a bell curve centred at 3.5.

**Why it is critical**

- It lets us use **normal-based methods** (confidence intervals, z-tests, t-tests) for means even when the population is not normal.
- It explains why sample means are reliable: the standard error σ / √n **shrinks as the sample size increases**, so larger samples give more precise estimates.
- It is the basis of much of **inferential statistics**, and is used in polling, quality control, A/B testing and forecasting.

**Conditions**

- The observations must be independent and randomly sampled.
- The population must have a finite mean and variance.
- The sample size must be sufficiently large.

## Question 7

> What is the significance of confidence intervals in statistical analysis?

### Answer

A **confidence interval (CI)** is a range of values, calculated from sample data, that is likely to contain the true population parameter (such as the mean). It has the form:

$$\text{point estimate} \;\pm\; (\text{critical value}) \times (\text{standard error})$$

For a mean:

$$\bar{x} \;\pm\; z \cdot \frac{s}{\sqrt{n}} \qquad (\text{or } t \text{ instead of } z)$$

**Example:** if the average delivery time in a sample is 30 minutes with a 95% CI of 28 to 32 minutes, we are fairly confident that the true average delivery time for all orders lies between 28 and 32 minutes.

**Significance**

- **Shows uncertainty:** a single sample mean is only an estimate. The CI shows how precise that estimate is, which a point estimate alone cannot do.
- **Width shows precision:** a narrow interval means a precise estimate; a wide one means more uncertainty. The width falls with a larger sample size or less variation in the data, and rises with a higher confidence level.
- **Supports decisions:** it shows the range of plausible values, so the practical importance of a result can be judged, not only whether it is "significant".
- **Links to hypothesis testing:** if a 95% CI does not contain the hypothesized value (for example 0 difference), the null hypothesis is rejected at the 5% significance level.
- **Comparing groups:** overlapping or non-overlapping intervals help compare means or proportions.

**Correct interpretation:** a 95% confidence level means that if the sampling were repeated many times, about 95% of the intervals built this way would contain the true parameter. It does not mean there is a 95% probability that one particular interval contains it.

## Question 8

> What is the concept of expected value in a probability distribution?

### Answer

The **expected value** E(X) of a random variable is its long-run average, the value we would expect on average if the random experiment were repeated a very large number of times. It is the probability-weighted mean of all possible values, and is the "centre of gravity" of the distribution.

Discrete:

$$E(X) = \sum x \cdot P(X = x) \qquad (\text{over all values } x)$$

Continuous:

$$E(X) = \int_{-\infty}^{\infty} x \, f(x) \, dx$$

**Examples**

- **Fair die:** E(X) = 1(1/6) + 2(1/6) + 3(1/6) + 4(1/6) + 5(1/6) + 6(1/6) = **3.5**. The expected value need not be a value that can actually occur.
- **A simple game:** win Rs. 100 with probability 0.2, lose Rs. 20 with probability 0.8. E(X) = 100(0.2) + (-20)(0.8) = 20 - 16 = **Rs. 4** per play, so the game is favourable in the long run.
- **Binomial distribution:** E(X) = n × p.

**Properties**

- E(aX + b) = a·E(X) + b.
- E(X + Y) = E(X) + E(Y) (always true, even if X and Y are dependent).

**Uses**

Expected value is used in decision making under uncertainty: insurance pricing, business profit forecasting, risk assessment, game and lottery analysis, and machine learning (expected loss).

## Question 9

> Write a Python program to generate 1000 random numbers from a normal distribution with mean = 50 and standard deviation = 5. Compute its mean and standard deviation using NumPy, and draw a histogram to visualize the distribution.

### Answer

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 1000 numbers from a normal distribution (mean 50, sd 5)
rng = np.random.default_rng(seed=42)   # seed makes results repeatable
data = rng.normal(loc=50, scale=5, size=1000)

# Mean and standard deviation using NumPy
print("Number of values             :", len(data))
print("Mean                         :", round(np.mean(data), 4))
print("Standard deviation (np.std)  :", round(np.std(data), 4))
print("Sample std deviation (ddof=1):", round(np.std(data, ddof=1), 4))
low, high = data.min(), data.max()
print("Minimum / Maximum            :", round(low, 2), "/", round(high, 2))

# Histogram of the generated data
plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, density=True, edgecolor="black",
         color="#9ecae1", label="Generated data")

# Theoretical normal curve N(50, 5) for comparison
x = np.linspace(data.min(), data.max(), 200)
pdf = np.exp(-0.5 * ((x - 50) / 5) ** 2) / (5 * np.sqrt(2 * np.pi))
plt.plot(x, pdf, color="red", linewidth=2, label="Normal(50, 5) curve")

plt.axvline(np.mean(data), color="black", linestyle="--",
            label=f"Sample mean = {np.mean(data):.2f}")
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Histogram of 1000 Random Numbers from a Normal Distribution")
plt.legend()
plt.show()
```

**Explanation:** `rng.normal(loc=50, scale=5, size=1000)` draws 1000 values from a normal distribution with mean 50 and standard deviation 5. The sample mean is **49.86** and the standard deviation is about **4.94** (4.95 when the sample formula with `ddof=1` is used). These are close to, but not exactly, the true values of 50 and 5, because a random sample never matches the population exactly. The histogram is bell-shaped and symmetric around 50, and follows the red normal curve closely. Almost all values lie between 35 and 65, which is within 3 standard deviations of the mean. The seed (42) makes the numbers repeatable; without it the values differ slightly on every run.

## Question 10

> You are working as a data analyst for a retail company. The company has collected daily sales data for 2 years and wants you to identify the overall sales trend.
>
> `daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255, 235, 260, 245, 250, 225, 270, 265, 255, 250, 260]`
>
> - Explain how you would apply the Central Limit Theorem to estimate the average sales with a 95% confidence interval.
> - Write the Python code to compute the mean sales and its confidence interval.

### Answer

**(a) Applying the Central Limit Theorem**

- Treat the daily sales values as a **random sample** from all the days of the business. The distribution of daily sales itself may not be normal.
- By the **CLT**, the sample mean is approximately normally distributed with mean μ and standard error σ / √n, so a normal-based confidence interval for the average sales can be built.
- **Step 1:** compute the sample mean x̄ and sample standard deviation s.
- **Step 2:** compute the standard error, SE = s / √n.
- **Step 3:** choose the critical value for 95% confidence: z = 1.96 with the CLT (large sample). Here n = 20 and σ is unknown, so the t-value with n - 1 = 19 degrees of freedom (2.093) is more accurate.
- **Step 4:** the 95% confidence interval is x̄ ± critical value × SE.
- **Caution:** n = 20 is below the usual guideline of 30 for the CLT, which is why the t-interval is preferred. Observations should be independent, and if sales show a clear trend or seasonality (higher on weekends, festival periods), the daily values are not identically distributed and a time-series method should be used to study the trend.

**(b) Python code and output**

```python
import numpy as np
from scipy import stats

daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

n = len(daily_sales)
mean_sales = np.mean(daily_sales)
std_sales = np.std(daily_sales, ddof=1)      # sample standard deviation
std_error = std_sales / np.sqrt(n)           # standard error of the mean

# 95% CI using the normal (z) critical value 1.96 (CLT approach)
z_crit = 1.96
ci_z = (mean_sales - z_crit * std_error, mean_sales + z_crit * std_error)

# 95% CI using the t-distribution (better for a small sample, sigma unknown)
t_crit = stats.t.ppf(0.975, df=n - 1)
ci_t = (mean_sales - t_crit * std_error, mean_sales + t_crit * std_error)

print("Sample size (n)         :", n)
print("Mean sales              :", round(mean_sales, 2))
print("Sample std deviation    :", round(std_sales, 2))
print("Standard error          :", round(std_error, 2))
print(f"95% CI (z = 1.96)       : ({ci_z[0]:.2f}, {ci_z[1]:.2f})")
print("t critical value (df=19):", round(t_crit, 3))
print(f"95% CI (t-distribution) : ({ci_t[0]:.2f}, {ci_t[1]:.2f})")
```

**Interpretation:** the mean daily sales are **248.25** with a sample standard deviation of 17.27 and a standard error of 3.86. The 95% confidence interval is **(240.17, 256.33)** using the t-distribution, or (240.68, 255.82) using z = 1.96. The t-interval is slightly wider, as expected for a small sample. We can say we are 95% confident that the true average daily sales lie between about **240 and 256** units.
